# Run SQL analysis files

This notebook runs each file in `sql/analysis/` and shows the query results one file at a time.

Use **Docker JupyterLab** (`http://localhost:8888`) so `DATABASE_URL` points at the `db` container.

```python
from pathlib import Path

from db_connect import get_database_url, get_engine
from sql_runner import run_sql_file

SQL_DIR = Path("../sql/analysis")

def show_sql_results(filename: str) -> None:
    # Create a fresh engine each run so an old @db URL cannot stick in memory.
    engine = get_engine()
    print(f"Connecting via: {get_database_url()}")
    path = SQL_DIR / filename
    print(f"Running: {path}")
    frames = run_sql_file(engine, path)
    print(f"Queries returned: {len(frames)} result table(s)\n")
    for i, df in enumerate(frames, start=1):
        print(f"--- Result {i} ---")
        display(df)
```

## 01 — Data validation
`sql/analysis/01_data_validation.sql`

```python
show_sql_results("01_data_validation.sql")
```

## 02 — Highest-demand tech occupations
`sql/analysis/02_highest_demand.sql`

```python
show_sql_results("02_highest_demand.sql")
```

## 03 — Highest-paying tech occupations

`sql/analysis/04_tech_concentration.sql`
